```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Data/Afficionado Coffee Roasters.xlsx - Transactions .csv")
```

```python
#Time based traansaction
```

```python
df["Hour"] = pd.to_datetime(df["transaction_time"]).dt.hour
```

```python

df["transaction_time"].head()
```

```python
df["Hour"] = pd.to_datetime(
    df["transaction_time"],
    format="%H:%M:%S"
).dt.hour
```

```python
df["Hour"].head()
```

```python
def time_of_day(hour):
    if 6 <= hour < 12:
        return "Morning"
    elif 12 <= hour < 17:
        return "Afternoon"
    elif 17 <= hour < 21:
        return "Evening"
    elif 21<= hour<= 24:
        return "Night"

df["Time_of_Day"] = df["Hour"].apply(time_of_day)
```

```python
df["Time_of_Day"].tail()
```

```python
 df["Revenue"] = df["transaction_qty"] * df["unit_price"]
```

```python
sns.histplot(df["Revenue"], bins=30)
plt.title("Revenue Distribution")
plt.show()
```

```python
hourly_sales = df.groupby("Hour")["Revenue"].sum()

hourly_sales.plot(kind="bar", figsize=(10,5))
plt.title("Revenue by Hour")
plt.xlabel("Hour")
plt.ylabel("Revenue")
plt.show()
```

```python
sns.countplot(data=df, x="Time_of_Day")
plt.title("Transactions by Time of Day")
plt.show()
```

```python
df.to_csv("Data/processed_data.csv", index=False)
```
