```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df=pd.read_csv("Data/Afficionado Coffee Roasters.xlsx - Transactions .csv")
```

```python
df.head()
```

```python

df.info()
```

```python
df.describe()
```

```python
df.isnull().sum()
```

```python
#Transaction quantity
```

```python

df["transaction_qty"].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["transaction_qty"], bins=20)
plt.title("Distribution of Transaction Quantity")
plt.xlabel("Quantity")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(6,3))
sns.boxplot(x=df["transaction_qty"])
plt.title("Transaction Quantity Boxplot")
plt.show()
```

```python
#customers usually purchase between 1 and 3 items per transaction. only few transactions contain unusually large quantities.
```

```python
#Unit price
```

```python
df["unit_price"].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["unit_price"], bins=20)
plt.title("Distibution of Unit price")
plt.xlabel("Price")
plt.ylabel("Frequency")
plt.show()
```

```python

sns.boxplot(x=df["unit_price"])
plt.title("Unit price Boxplot")
plt.show()
```

```python
#Product usually cost moderate price while some products are highly priced 
```

```python
#Revenue
```

```python
df["Revenue"] = df["transaction_qty"] * df["unit_price"]
```

```python
df["Revenue"].describe()
```

```python
plt.figure(figsize=(8,5))
plt.hist(df["Revenue"], bins=20)
plt.title("Distribution of Revenue")
plt.xlabel("Revenue")
plt.ylabel("Frequency")
plt.show()
```

```python

sns.boxplot(x=df["Revenue"])
plt.title("Revenue Boxplot")
plt.show()
```

```python
#Most products do not generate much revenue while small number or high valued products generate high revenue
```

```python
#product cateegory
```

```python
df["product_category"].value_counts()
```

```python
plt.figure(figsize=(20,5))
plt.hist(df["product_category"], bins=20)
plt.title("Transaction by product category")
plt.xticks(rotation=45)
plt.xlabel("Product category")
plt.ylabel("Transaction")
plt.show()
```

```python
#Coffe is the most purchased product category
```

```python
##Store Location
```

```python
plt.figure(figsize=(8,4))
plt.hist(df["store_location"], bins=5)
plt.xticks(rotation=45)
plt.title("Transaction by Store location")
plt.show()
```

```python
df["store_location"].value_counts()
```

```python
#Hell's kitchen sells the most
```

```python
#product type
```

```python
plt.figure(figsize=(15,6))
sns.countplot(data=df, y="product_type", order=df["product_type"].value_counts().index)
plt.title("Product Type Distribution")
plt.show()
```

```python
#Brewed chai tea is one of the most sold product type
```
